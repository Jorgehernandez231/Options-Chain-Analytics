```python
# !/usr/bin/env python3
# Script to fetch and display top-level keys and data structure from a JSON API response

import requests, json
u = "https://cdn.cboe.com/api/global/delayed_quotes/options/_SPX.json"
j = requests.get(u, timeout=45).json()
print("TOP:", list(j.keys()))
print("DATA KEYS:", list(j.get("data", {}).keys()))
print("LEN options:", len(j.get("data", {}).get("options", [])))
print("FIRST OPTION OBJECT:")
print(json.dumps(j["data"]["options"][0], indent=2)[:1500])
```

```python
# Count how many option objects have call and put data fields populated 
opts = j["data"]["options"]
print(
    sum(1 for o in opts if isinstance(o.get("call"), dict)),
    sum(1 for o in opts if isinstance(o.get("put"), dict))
)
```
